```python
# this are accesible via the base_parameters in the yml file
catalog = dbutils.widgets.get("catalog")
pipeline_schema = dbutils.widgets.get("pipeline_schema")

full_schema = f"{catalog}.{pipeline_schema}"
```

```python
# Checks whether a table exists or not
def table_exists(table_name, path):
    if spark.catalog.tableExists(f"{path}.{table_name}"):
        return True

    else:
        return False
```

```python
bronze_tables = [
    "bronze_customers",
    "bronze_telematics",
    "bronze_claims",
    "bronze_policies",
    "bronze_payments",
    "bronze_agents",
]

silver_tables = [
    "silver_customers",
    "silver_customers_quarantine",
    "silver_policies",
    "silver_policies_quarantine",
    "silver_policy_coverages",
    "silver_claims",
    "silver_claims_quarantine",
    "silver_payments",
    "silver_payments_quarantine",
    "silver_telematics",
    "silver_telematics_quarantine",
    "silver_agents",
]

gold_tables = [
    "dim_date",
    "dim_policy",
    "dim_customer",
    "fact_premium",
    "fact_payments",
    "fact_claims",
    "agg_agent_performance",
    "agg_claims_monthly",
    "agg_customer_value",
    "agg_loss_ratio",
    "agg_telematics_risk",
]

all_layers = [bronze_tables, silver_tables, gold_tables]
```

```python
# Checks whether all necessary tables exist
def all_tables_exist(status):
    for layer in all_layers:
        for table in layer:
            if not table_exists(table, full_schema):
                status = "FAIL"
                return status

    return status
```

```python
status = all_tables_exist(status)
```

```python
# this is for giving the tasks of a job access to my variable status
dbutils.jobs.taskValues.set(key="status", value=status)
```
